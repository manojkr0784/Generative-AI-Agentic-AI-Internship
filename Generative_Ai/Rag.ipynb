```python
loader=UnstructuredFileLoader("HRPolicy.pdf")
```

```python
text_splitter=CharacterTextSplitter(
    chunk_size=1000, chunk_overlap=300
    )
```

```python
document=loader.load()
texts=text_splitter.split_documents(document)
```

```python
embeddinds=HuggingFaceEmbeddings()
```

```python
persisit_directory="vector_db"
```

```python
vectordb=Chroma.from_documents(
    embedding=embeddinds,
    documents=texts,
    persist_directory=persisit_directory)
```

```python
retriever=vectordb.as_retriever()
```

```python
llm=ChatGroq(model="llama-3.3-70b-versatile", temperature=0)
```

```python
qa_chain=RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=retriever,
    return_source_documents=True
)
```

```python
query="How many leaves are allowed annually"
response=qa_chain.invoke({"query":query})
```
