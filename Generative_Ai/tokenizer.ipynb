Assignment 3: Find the longest token in the sentence. Machine Learning is amazing.

```python
text = "Machine Learning is Amazing"
tokens = text.split()
longest = max(tokens, key=len)
print(longest)
```

Assignment 5 : Count frequency of every token.

Assignment 7 : Question - Create vocabulary from text.

```python
text = "I love AI and NLP"
tokens = text.split()
vocab = sorted(set(tokens))
print(vocab)
```

Assignment 8 : Convert tokens into token IDs.

```python
text = "I love AI"
tokens = text.split()
vocab = {word:i for i, word in enumerate(set(tokens))}
token_ids = [vocab[word] for word in tokens]
print(token_ids)
```

Assignment 9 : Decode token IDs back into text.

```python
vocab = { 
    "I":0, 
    "love":1, 
    "AI":2 
} 
reverse_vocab = {v:k for k,v in vocab.items()} 
ids = [0,1,2] 
sentence = [reverse_vocab[i] for i in ids] 
print(" ".join(sentence))
```
